
# Laboratory 1A — From a CIF file to a crystal structure

In this guided activity we use **MgO (periclase)** as a deliberately simple
reference crystal. The aim is not to learn Python syntax. The code is already
prepared so that we can concentrate on the crystallography.

We will ask four connected questions:

1. What information is actually stored in a CIF file?
2. How does the asymmetric unit become a complete unit cell?
3. Can the symmetry be reconstructed from the lattice and atomic positions?
4. How are Wyckoff positions related to the atomic sites we see in the structure?

The notebook uses **Gemmi** to read crystallographic information,
**spglib** to determine symmetry numerically, and **py3Dmol** for interactive
visual inspection.


```python

from pathlib import Path

from mineral_science_tools import (
    asymmetric_unit_table,
    read_cif,
    show_structure,
    structure_summary,
    symmetry_generators,
    symmetry_operations,
    symmetry_summary,
    unit_cell_table,
    wyckoff_table,
)

CIF_FILE = Path("data/MgO.cif")
```


## 1. Read the CIF

A CIF is not a picture of a crystal. It is a structured description containing
quantities such as unit-cell parameters, symmetry information, atomic labels,
fractional coordinates and occupancies.

Gemmi reads these data and creates an internal crystallographic object.


```python

structure = read_cif(CIF_FILE)
structure_summary(CIF_FILE)
```

## 2. The asymmetric unit is not the complete unit cell

The CIF lists the symmetrically independent sites needed to describe the
structure. Applying the space-group operations generates the remaining
symmetry-equivalent positions in the conventional unit cell.

First inspect the sites written explicitly in the CIF.


```python

asymmetric_unit_table(CIF_FILE)
```


Now expand them using the space-group symmetry.


```python

unit_cell_table(CIF_FILE)
```

## 4. Symmetry: what the CIF says and what the coordinates imply

There are two different operations taking place here.

**Gemmi** reads the crystallographic description supplied by the CIF, including
its space group.

**spglib** is given the lattice vectors, atomic positions and chemical species
and searches for operations that map the structure onto itself within a
numerical tolerance.

Agreement between the two is therefore a useful consistency check.


```python

symmetry_summary(CIF_FILE, symprec=1e-5)
```

## 5. A space group is a set of operations

For a high-symmetry crystal it is not useful to print hundreds of operations
without structure. Gemmi can instead expose the operations encoded by the Hall
symbol together with the centring translations from which the full group is
generated.

The coordinate triplet `x,y,z` is the identity. Expressions such as
`-y,x,z` describe how fractional coordinates are transformed.


```python

gens = symmetry_generators(CIF_FILE)
print("Hall-derived operations")
display(gens["operations"])

print("Centring translations")
display(gens["centring"])
```


For comparison, inspect only the first few operations of the complete space
group.


```python

symmetry_operations(CIF_FILE, limit=12)
```

## 6. Wyckoff positions

A Wyckoff position collects sites related by the symmetry of a space group.
Its **multiplicity** tells us how many equivalent sites of that orbit occur in
the conventional unit cell; the **Wyckoff letter** distinguishes inequivalent
orbits; and the **site-symmetry symbol** describes the symmetry that leaves the
site itself invariant.


```python

wyckoff_table(CIF_FILE)
```
