## 3. Inspect one output manually

Before asking Python to extract anything, open `outputs/mgo_shrink2.out` as a
text file.

Find:

- `SHRINK. FACT.(MONKH.)`;
- `NUMBER OF K POINTS IN THE IBZ`;
- several lines beginning with `CYC`;
- `SCF ENDED - CONVERGENCE ON ENERGY`;
- the final `END ... TELAPSE ... TCPU ...` line;
- the normal `TERMINATION` marker.

These are the pieces of information that the parser will later use.

````python

from pathlib import Path
from mineral_science_tools import parse_crystal_scf_output

OUTPUT = Path("outputs/mgo_shrink2.out")
result = parse_crystal_scf_output(OUTPUT)
result
````


The parser is intentionally conservative. If the SCF-convergence marker or
normal termination is absent, it raises an error instead of returning an energy.


````python

print(f"SHRINK:              {result.shrink}")
print(f"k-points in IBZ:     {result.kpoints_ibz}")
print(f"Energy (Ha):          {result.energy_ha:.12f}")
print(f"SCF cycles:           {result.scf_cycles}")
print(f"Elapsed time (s):     {result.elapsed_s:.2f}")
print(f"CPU time (s):         {result.cpu_s:.2f}")
````
