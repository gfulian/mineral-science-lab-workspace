## 3. Convert the CRYSTAL result to a Quantas phonon input

## 4. Why Γ is important for spectroscopy

For first-order IR and Raman scattering, the photon momentum is very small on
the scale of the Brillouin zone. The measured optical phonons are therefore
near $\mathbf q=0$.

The $\Gamma$ calculation is consequently valuable spectroscopic information
even though it is a poor sampling of the full phonon spectrum needed for bulk
thermodynamics.

## 5. Calculate harmonic heat capacity with Quantas

We now deliberately use the incomplete $\Gamma$ spectrum to calculate a
harmonic $C_V(T)$ curve.


````python

cv_gamma = ha_cv_curve(
    YAML,
    temperature_min=0.0,
    temperature_max=1200.0,
    temperature_step=10.0,
)

cv_gamma.head()
````

````python

experimental = load_experimental_heat_capacity(
    "data/mgo_experimental_heat_capacity.csv"
)

plot_cv_comparison(
    {"Gamma only": cv_gamma},
    experimental,
);
````
